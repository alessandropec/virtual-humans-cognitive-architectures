# Module 0: direct LLM calls

Open this notebook from its `Code` folder and select the course Python kernel.
Run the key-loading cell, then the provider you want to try.

Copy `virtual_human_course/llm_provider_api_key.example.yaml` to
`llm_provider_api_key.yaml` in the same course folder and add your keys.

````python
from pathlib import Path
import yaml

key_file = Path("../../../llm_provider_api_key.yaml")
provider_keys = yaml.safe_load(key_file.read_text())

GEMINI_API_KEY = provider_keys["gemini"]["api_key"]
GROQ_API_KEY = provider_keys["groq"]["api_key"]
OPENROUTER_API_KEY = provider_keys["openrouter"]["api_key"]
OLLAMA_API_KEY = provider_keys["ollama"]["api_key"]
````

## 1a. Ollama local: optional small-model check

### Before running the local cells

1. Download and install the **Ollama application** for your operating system from [ollama.com/download](https://ollama.com/download).
2. Start Ollama and keep it running. On Linux, run `ollama serve` in a separate terminal if the service is not already running. The local API normally uses `http://localhost:11434`.
3. In the VS Code terminal, download the small model used below:

   ```sh
   ollama pull llama3.2:1b
   ```

4. Select the shared `virtual_human_project/.venv-vh-course` kernel in this notebook. The course `requirements.txt` installs the `ollama` Python package.

The Python package is the client library: installing it does **not** install the Ollama application or download model weights. The model needs disk space and enough memory to run on your computer. Local calls require no account or API key.

The next cell lists the models installed locally. Download the model first, then run that cell and the response cell below.

For **direct Ollama Cloud calls**, skip the application installation and model download. Use the cloud section with your account and API key instead.


````python
from ollama import Client

client = Client(host="http://localhost:11434")

for model in client.list().models:
    print(model.model)
````

Choose a model name from the list above and set `model=` in the next cell. The cloud client lists cloud models; the local client lists models installed on this computer.


````python
response = client.chat(
    model="llama3.2:1b",
    messages=[{"role": "user", "content": "Say hello!"}],
)
print(response.message.content)
````

## 1b. Ollama Cloud

Create an account and an [API key](https://ollama.com/settings/keys) and add it to `virtual_human_course/llm_provider_api_key.yaml`. Run the initial key-loading cell. Choose a model from the list below and set `model=` in the response cell. No local Ollama application or model download is required.


````python
from ollama import Client

client = Client(
    host="https://ollama.com",
    headers={"Authorization": f"Bearer {OLLAMA_API_KEY}"},
)

for model in client.list().models:
    print(model.model)
````

Choose a model name from the list above and set `model=` in the next cell. The cloud client lists cloud models; the local client lists models installed on this computer.


````python
response = client.chat(
    model="gpt-oss:120b",
    messages=[{"role": "user", "content": "Say hello!"}],
)
print(response.message.content)
````

## 2. Google Gemini

Create a key in [Google AI Studio](https://aistudio.google.com/). The key comes from the initial YAML-loading cell. Set the model directly in the code. Choose an available model from your account.


````python
from google import genai

client = genai.Client(api_key=GEMINI_API_KEY)
````

### Models for this exercise

Model IDs as of **16 October 2026**:

- `gemini-3.5-flash-lite`
- `gemini-3.6-flash`

Set `model=` to either ID in the next cell. If a model is unavailable, try the other. Access can depend on your account and quota, and model IDs can change or become deprecated.

Current information: [Google model catalogue](https://ai.google.dev/gemini-api/docs/models), [deprecations](https://ai.google.dev/gemini-api/docs/deprecations), [AI Studio usage and limits](https://aistudio.google.com/usage).


````python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="Say hello!",
)
print(response.text)
````

## 3. Groq

Create an account and API key in the [Groq console](https://console.groq.com/) and add your key to `virtual_human_course/llm_provider_api_key.yaml`. Run the initial key-loading cell, then create the client below.


````python
from groq import Groq

client = Groq(api_key=GROQ_API_KEY)
````

### Models for this exercise

Model IDs as of **16 October 2026**:

- `openai/gpt-oss-20b`
- `openai/gpt-oss-120b`

Set `model=` to either ID in the next cell. If a request fails, check your free-plan quota and whether the model is still available.

Current information: [Groq supported models](https://console.groq.com/docs/models), [deprecations](https://console.groq.com/docs/deprecations), [rate limits](https://console.groq.com/docs/rate-limits).


````python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[{"role": "user", "content": "Say hello!"}],
)
print(response.choices[0].message.content)
````

## 4. OpenRouter

Use your [OpenRouter](https://openrouter.ai/) key from the initial cell and set the model ID. OpenRouter accepts an OpenAI-compatible client: the `base_url` points to OpenRouter.


````python
from openai import OpenAI

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY,
)
````

### Free models for this exercise

Free model IDs as of **16 October 2026**:

- `liquid/lfm-2.5-2.6b:free`
- `google/gemma-4-26b-a4b-it:free`

Start with `liquid/lfm-2.5-2.6b:free`. Keep the `:free` suffix when setting `model=`. Free models have usage limits and can become unavailable or be retired. If you receive HTTP 429, check your quota or try again later. You can also choose another free text model from the catalogue.

Current information: [free-model collection](https://openrouter.ai/collections/free-models), [model catalogue](https://openrouter.ai/models), [free variants](https://openrouter.ai/docs/guides/routing/model-variants/free), [rate limits](https://openrouter.ai/docs/api-reference/limits). Check that both input and output prices are zero.


````python
response = client.chat.completions.create(
    model="liquid/lfm-2.5-2.6b:free",
    messages=[{"role": "user", "content": "Say hello!"}],
)
print(response.choices[0].message.content)
````
